## Learning Wave 1 — The String Toolkit

### 🧠 First: what *is* a string, really?

A **string** is just a **row of characters** — exactly like a list, but every slot holds one character. So everything you already know about lists still works:

- `text[0]` is the first character, `text[-1]` the last
- `len(text)` is how many characters there are
- `"bike" in text` checks whether that word appears

What strings add is a **toolbox of methods** for cleaning and reshaping text. The ones you will use constantly:

| Method | What it does |
|---|---|
| `.strip()` | remove spaces / newlines at the start and end |
| `.lower()` / `.upper()` | change case, so `"Bike"` and `"bike"` count as the same word |
| `.replace(old, new)` | swap one piece of text for another |
| `.split()` | cut a sentence into a **list of words** |
| `" ".join(words)` | glue a list of words back into one sentence |
| `.count("x")` | how many times `"x"` appears |

**The big idea for the whole week:** a computer cannot understand a sentence. The first move in *every* text and AI pipeline is to **clean** the text and **split it into pieces you can count.**

## Learning Wave 2 — From Text to Numbers

One response is easy. But a city gets *thousands*. To find what people care about, you count words across **all** of them — this is called a **bag of words**, and it is the foundation of almost all classic NLP.

The recipe reuses exactly what you already know:
1. `for` over every response (Week 2)
2. `.lower().split()` each response into words (Week 4)
3. the **counting pattern** from Week 3: `counts[word] = counts.get(word, 0) + 1`

### Exercise 2: Which words come up most?
⏱ **15 minutes** | Slide 10

**Given:**
```python
responses = [
    "The bike lanes are great but too narrow",
    "More bike parking please the parking is always full",
    "Traffic lights are too short for pedestrians",
    "Great new park but we need more lights at night",
    "The park is lovely and the bike lanes are a great idea",
]
```

**Steps:**
1. Build a dictionary `counts` of how often each (lowercased) word appears, using the counting pattern.
2. Print every word with its count using `.items()`.
3. Find the **most mentioned** word with the comparison-variable pattern (Weeks 2–3 — no need for `max()` on a dict).
4. **Bonus:** make a `stopwords` list (`"the"`, `"are"`, `"but"`, `"and"`, `"a"`, `"is"`, `"for"`, …) and skip those words with `continue`, so only meaningful words remain.

````python
# Exercise 2: Which words come up most?
responses = [
    "The bike lanes are great but too narrow",
    "More bike parking please the parking is always full",
    "Traffic lights are too short for pedestrians",
    "Great new park but we need more lights at night",
    "The park is lovely and the bike lanes are a great idea",
]

# 1. Count every word
counts = {}
for response in responses:
    for word in response.lower().split():
        counts[word] = counts.get(word, 0) + 1

# 2. Print every word and its count
for word, count in counts.items():
    print(word, count)

# 3. Find the most mentioned word without max() on a dict
most_word = None
most_count = 0
for word, count in counts.items():
    if count > most_count:
        most_word = word
        most_count = count

print("Most mentioned word:", most_word)
print("Count:", most_count)

# 4. Bonus: skip stopwords
stopwords = {"the", "are", "but", "and", "a", "is", "for", "more", "we", "to", "at"}
meaningful_counts = {}

for response in responses:
    for word in response.lower().split():
        if word in stopwords:
            continue
        meaningful_counts[word] = meaningful_counts.get(word, 0) + 1

print("\nMeaningful words:")
for word, count in meaningful_counts.items():
    print(word, count)
````

## Learning Wave 3 — Reading & Writing Files

So far the data lived *inside* your code. Real data lives in **files**. Python opens files with `open()`, and the safe way is the `with` block — it closes the file for you automatically:

```python
# Writing
with open("notes.txt", "w", encoding="utf-8") as f:
    f.write("first line\n")      # \n = a new line

# Reading, line by line
with open("notes.txt", "r", encoding="utf-8") as f:
    for line in f:
        print(line.strip())      # strip() removes the trailing newline
```

- `"w"` = write (overwrites the whole file), `"a"` = append, `"r"` = read.
- Always pass `encoding="utf-8"` so accents and symbols don't break.

### Exercise 3: Read feedback from a file
⏱ **20 minutes** (incl. reflection) | Slide 13

**Steps:**
1. Write the lines below to a file `feedback.txt` (use `writelines()`; each line already ends in `\n`).
2. Open `feedback.txt` and loop over it line by line. `strip()` each line; if the line is empty, `continue`.
3. Count the **total** responses, and count how many contain the word `"ALERT"`. Print each urgent line.
4. Write a short report to `report.txt`: total responses and number of urgent ones.
5. **Reflection** (markdown cell below).

```python
lines = [
    "The bike lanes are great\n",
    "ALERT broken glass on the path near the school\n",
    "More lights please it is too dark at night\n",
    "\n",
    "ALERT the crossing feels unsafe for children\n",
    "Lovely new park thank you\n",
```

````python
# Exercise 3: Read feedback from a file

# 1. Write the data file
lines = [
    "The bike lanes are great\n",
    "ALERT broken glass on the path near the school\n",
    "More lights please it is too dark at night\n",
    "\n",
    "ALERT the crossing feels unsafe for children\n",
    "Lovely new park thank you\n",
]

with open("feedback.txt", "w", encoding="utf-8") as f:
    f.writelines(lines)

# 2 & 3. Read it back, skip empty lines, count total + urgent, print urgent lines
total = 0
urgent = 0

with open("feedback.txt", "r", encoding="utf-8") as f:
    for line in f:
        cleaned_line = line.strip()

        # Skip empty lines
        if not cleaned_line:
            continue

        total += 1

        # Check for urgent items
        if "ALERT" in cleaned_line:
            urgent += 1
            print("Urgent feedback:", cleaned_line)

print("Total responses:", total)
print("Urgent responses:", urgent)

# 4. Write a small report to report.txt
with open("report.txt", "w", encoding="utf-8") as f:
    f.write(f"Total responses: {total}\n")
    f.write(f"Urgent responses: {urgent}\n")
````

````python
# Big Assignment: Community Feedback Analyzer
# Design your own solution. It must: read a file, clean + split + count the text,
# and write a report. Build and test it one piece at a time.
import re

# 1. STOPWORDS & URGENT/SENTIMENT LISTS
STOPWORDS = {
    "the", "a", "an", "and", "or", "but", "in", "on", "at", "to", "for",
    "of", "with", "is", "are", "was", "were", "it", "this", "that", "i",
    "you", "we", "they", "my", "our", "new", "more", "please", "too", "very"
}

URGENT_KEYWORDS = {"unsafe", "broken", "alert", "danger", "dark"}
POSITIVE_WORDS = {"great", "lovely", "thank", "good", "nice", "enjoy"}
NEGATIVE_WORDS = {"dark", "unsafe", "broken", "bad", "terrible", "dirty"}

# 2. HELPER FUNCTIONS

def read_responses(filepath):
    """Reads lines from a file and skips blank entries."""
    responses = []
    with open(filepath, "r", encoding="utf-8") as f:
        for line in f:
            cleaned = line.strip()
            if cleaned:
                responses.append(cleaned)
    return responses

def clean_and_tokenize(text):
    """Cleans a string and returns a list of individual words."""
    cleaned = text.lower()
    for char in ".,!?\"'-":
        cleaned = cleaned.replace(char, "")
    return cleaned.split()

def analyze_feedback(responses):
    """Extracts statistics, word frequencies, urgent messages, and sentiment."""
    total_responses = len(responses)
    total_words = 0
    word_counts = {}
    urgent_responses = []
    positive_count = 0
    negative_count = 0
    numbers_found = []

    for resp in responses:
        words = clean_and_tokenize(resp)
        total_words += len(words)

        # Check for urgent items
        if any(keyword in words for keyword in URGENT_KEYWORDS):
            urgent_responses.append(resp)

        # Count frequencies & sentiment
        for word in words:
            if word in POSITIVE_WORDS:
                positive_count += 1
            if word in NEGATIVE_WORDS:
                negative_count += 1

            if word not in STOPWORDS:
                word_counts[word] = word_counts.get(word, 0) + 1

        # Bonus: Extract numbers mentioned
        found_nums = re.findall(r"\d+", resp)
        if found_nums:
            numbers_found.extend(found_nums)

    # Calculate top 5 keywords
    sorted_words = sorted(word_counts.items(), key=lambda item: item[1], reverse=True)
    top_5_keywords = sorted_words[:5]

    avg_words = total_words / total_responses if total_responses > 0 else 0

    return {
        "total_responses": total_responses,
        "avg_words": avg_words,
        "top_5": top_5_keywords,
        "urgent_responses": urgent_responses,
        "positive_count": positive_count,
        "negative_count": negative_count,
        "numbers_found": numbers_found
    }

def generate_report(stats):
    """Formats the statistical results into a readable text report."""
    report_lines = []
    report_lines.append("==========================================")
    report_lines.append("       COMMUNITY FEEDBACK REPORT          ")
    report_lines.append("==========================================\n")

    report_lines.append(f"Total Responses Analyzed: {stats['total_responses']}")
    report_lines.append(f"Average Words per Response: {stats['avg_words']:.2f}\n")

    report_lines.append("Top 5 Keywords (excluding stopwords):")
    for word, count in stats["top_5"]:
        report_lines.append(f"  - {word}: {count} times")

    report_lines.append(f"\nUrgent Responses Detected ({len(stats['urgent_responses'])}):")
    for item in stats["urgent_responses"]:
        report_lines.append(f"  - ALERT: \"{item}\"")

    report_lines.append(f"\nSentiment Breakdown:")
    report_lines.append(f"  - Positive Words: {stats['positive_count']}")
    report_lines.append(f"  - Negative Words: {stats['negative_count']}")

    if stats["numbers_found"]:
        report_lines.append(f"\nSpecific Numbers Mentioned: {', '.join(stats['numbers_found'])}")

    report_lines.append("\n==========================================")
    return "\n".join(report_lines)

# 3. MAIN PIPELINE EXECUTION

# Step A: Load data
file_path = "feedback.txt"
responses = read_responses(file_path)

# Step B: Process data
stats = analyze_feedback(responses)
report_text = generate_report(stats)

# Step C: Print report to console
print(report_text)

# Step D: Save report to file
with open("feedback_report.txt", "w", encoding="utf-8") as f:
    f.write(report_text)

# Step E: Bonus - Interactive Keyword Search
search_term = input("\nSearch for a keyword in feedback (or press Enter to skip): ").strip().lower()
if search_term:
    matches = [r for r in responses if search_term in clean_and_tokenize(r)]
    print(f"Found {len(matches)} response(s) containing '{search_term}':")
    for m in matches:
        print(f"  -> {m}")
````

## 5 Things to Remember

1. **A string is a row of characters** — indexing, slicing, `len()` and `in` work just like with lists.
2. **Clean before you count** — `.strip()`, `.lower()` and `.replace()` turn messy text into something comparable (`"Bike!"` and `"bike"` become the same).
3. **`.split()` turns a sentence into a list of words; `" ".join()` turns it back** — this is *tokenisation*, the first step of every NLP system.
4. **Counting words = the Week 3 counting pattern** — `counts[word] = counts.get(word, 0) + 1` is a "bag of words".
5. **Files in, results out** — `with open(...) as f:` reads real data and writes reports safely. And remember: *the words missing from the file are missing from everything you build on it.*
